```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
df = pd.DataFrame(columns=["Model","Data", "Odds Ratio", "PPV", "k", "Total", "Positive", "Negative"])

for file in os.listdir("./results/"):
    if not file.endswith(".txt") or not file.startswith("BerryFinder_"):
        continue
    file_name = file.split(".txt")[0]
    file_args = file_name.split("_")
    data = file_args[1]
    params = file_args[2].split("-")
    odds_ratio = float(params[0])
    ppv = float(params[1])
    file_path = f"./results/{file}"
    # A 1% coverage threshold was applied for the MIMIC-IV dataset
    if data == "MIMIC-IV":
        file_path = f"./results_BF_low_coverage/{file}"
    with open(file_path, "r") as f:
        lines = f.readlines()
        total = float(lines[0].split(": ")[1])
        positive = float(lines[1].split(": ")[1])
        negative = float(lines[2].split(": ")[1])
        k = int(lines[3].split(": ")[1])
        df.loc[len(df)] = ["BerryFinder", data, odds_ratio, ppv, k, total, positive, negative]
```

```python
df["Youden's J Index"] = df["Positive"] - df["Negative"]
```

```python
# df.groupby("Data").agg({"Jaccard": "max"})
# Get k for max Jaccard per Data
df.loc[df.groupby("Data")["Youden's J Index"].idxmax()][["Data", "k"]].set_index("Data").to_dict()["k"]
```

```python
for d in df["Data"].unique():
    df_d = df[df["Data"] == d]
    df_d = df_d.sort_values(by=["Youden's J Index"], ascending=False)
    df_d[["Odds Ratio", "PPV", "Youden's J Index"]].to_csv("berry_results_csv/"+d+".csv", index=False)
    display(f"Top 10 results for {d} dataset")
    display(df_d.head(10))
```
