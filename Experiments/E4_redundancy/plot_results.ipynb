```python
import pandas as pd
import os 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def get_refinements(df):
    df.Description = df.Description.apply(lambda x: set(x.split(", ")))
    refinements = 0
    for i,row1 in df.iterrows():
        for j, row2 in df.iterrows():
            p1 = row1.Description
            p2 = row2.Description
            q1 = row1.Quality
            q2 = row2.Quality
            # Refinement condition
            if p1 < p2 and q1 >= q2:
                refinements += 1
                # print(f"Refinement found: {p1} -> {p2} with qualities {q1} -> {q2}")
    return refinements
```

```python
df = pd.DataFrame(columns=["Dataset", "Model", "Refinements"])
for file in os.listdir("./results"):
    if file.endswith(".csv"):
        file_name = file.split(".csv")[0]
        dataset, model = file_name.split("_")
        subgroups = pd.read_csv(os.path.join("./results", file))
        refinements = get_refinements(subgroups)
        df.loc[len(df)] = [dataset, model, refinements]
```
