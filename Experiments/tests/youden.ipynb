```python
import pandas as pd
```

```python
df = pd.read_csv("../data/vancomycin.csv")
```

```python
res = pd.read_csv("./results/BerryFinder_Vancomycin.csv", index_col=0)
```

```python
res["Description"]
```

```python
target = df["culture_susceptibility"] == "R"
```

```python
global_entry = pd.Series(False, index = df.index)
for i, row in res.iterrows():
    entry = pd.Series(True, index = df.index)
    desc = row["Description"].split(", ")
    for d in desc:
        column = d.split(" = ")[0]
        value = d.split(" = ")[1]
        value = value.replace("'", "")
        entry &= df[column] == value
    global_entry |= entry
```
