```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../dermatology_original.csv")
df.head()
```

```python
# Move class column to the beginning
class_col = df.pop('class')
df.insert(0, 'class', class_col)
df.head()
```

```python
# Check type and values of each column
for c in df.columns:
    print("Column:", c)
    print("Type:", df[c].dtype)
    print("Number of unique values:", df[c].nunique())
```

```python
df["class"].value_counts()
```

```python
# Age is a numeric column, remove it
df = df.drop(columns=['age'])
# Cast columns to str type for subgroup library
for c in df.columns:
    df[c] = df[c].astype(str)
# Save the processed DataFrame
df.to_csv("../dermatology.csv", index=False)
```

```python
df_train, df_test = train_test_split(df, test_size=0.2, stratify=df["class"], random_state=42)
df_tuning, df_validation = train_test_split(df_train, test_size=0.2, stratify=df_train["class"], random_state=42)
df_train.to_csv("../dermatology_train.csv", index=False)
df_test.to_csv("../dermatology_test.csv", index=False)
df_tuning.to_csv("../dermatology_tuning.csv", index=False)
df_validation.to_csv("../dermatology_validation.csv", index=False)
```
