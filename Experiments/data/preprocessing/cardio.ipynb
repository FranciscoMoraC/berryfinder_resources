```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../cardio_original.csv', sep = ";")
print(df.shape)
df.head()
```

```python
# Remove "id" column
df = df.drop(columns=['id'])
# Move class column to the beginning
class_col = df.pop('cardio')
df.insert(0, 'cardio', class_col)
df.head()
```

```python
# Check type and values of each column
for c in df.columns:
    print("Column:", c)
    print("Type:", df[c].dtype)
    print("Number of unique values:", df[c].nunique())
```

```python
df.cardio.value_counts()
```

```python
# Remove "age", "height", "weight", "ap_hi", and "ap_lo" columns
df = df.drop(columns=['age', 'height', 'weight', 'ap_hi', 'ap_lo'])
# Cast columns to str type for subgroup library
for c in df.columns:
    df[c] = df[c].astype(str)
# Save the processed DataFrame
df.to_csv("../cardio.csv", index=False)
```

```python
df_train, df_test = train_test_split(df, test_size=0.2, random_state=42, stratify=df["cardio"])
df_tuning, df_val = train_test_split(df_train, test_size=0.2, random_state=42, stratify=df_train["cardio"])
df_train.to_csv("../cardio_train.csv", index=False)
df_test.to_csv("../cardio_test.csv", index=False)
df_tuning.to_csv("../cardio_tuning.csv", index=False)
df_val.to_csv("../cardio_validation.csv", index=False)
```
