```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("../depression_original.csv")
print(df.shape)
df.head()
```

```python
# Move class column to the beginning
class_col = df.pop('Depression')
df.insert(0, 'Depression', class_col)
df.head()
```

```python
# Check type and values of each column
for c in df.columns:
    print("Column:", c)
    print("Type:", df[c].dtype)
    print("Number of unique values:", df[c].nunique())
```

```python
df.Depression.value_counts()
```

```python
#Remove white spaces and '?' from column names
df.columns = df.columns.str.replace('?', '', regex=False)
df.columns = df.columns.str.replace(' ', '_', regex=False)
df.columns
```

```python
# Remove 'Age' and 'Study_Hours' columns
df = df.drop(columns=['Age', 'Study_Hours'])
# Cast columns to str type for subgroup library
for c in df.columns:
    df[c] = df[c].astype(str)
# Save the processed DataFrame
df.to_csv("../depression.csv", index=False)
```

```python
df_train, df_test = train_test_split(df, test_size=0.2, stratify=df["Depression"], random_state=42)
df_tuning, df_validation = train_test_split(df_train, test_size=0.2, stratify=df_train["Depression"], random_state=42)
df_train.to_csv("../depression_train.csv", index=False)
df_test.to_csv("../depression_test.csv", index=False)
df_tuning.to_csv("../depression_tuning.csv", index=False)
df_validation.to_csv("../depression_validation.csv", index=False)
```
