```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('../healthy_aging_original.csv')
# Replace blank spaces in column names
df.columns = df.columns.str.replace(' ', '_')
df = df.astype("str")
# Remove Age column (only one value)
df.drop(columns=["Age"], inplace=True)
```

```python
df_train, df_test = train_test_split(df, test_size=0.2, stratify=df["Number_of_Doctors_Visited"], random_state=42)
df_tuning, df_validation = train_test_split(df_train, test_size=0.2, stratify=df_train["Number_of_Doctors_Visited"], random_state=42)
```

```python
df_train.to_csv("../healthy_aging_train.csv", index=False)
df_test.to_csv("../healthy_aging_test.csv", index=False)
df_tuning.to_csv("../healthy_aging_tuning.csv", index=False)
df_validation.to_csv("../healthy_aging_validation.csv", index=False)
df.to_csv("../healthy_aging.csv", index=False)
```
